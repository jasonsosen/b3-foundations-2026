## Q8｜内積を、手計算とコードで比べる［15分］

### 完成例：一定の力がした仕事

力`F=[2, 1]` N、変位`s=[3, 0]` mとします。一定の力に対する仕事は **W = F・s = 2×3 + 1×0 = 6 J** です。
ここでの二つの成分は同じ直交座標系のx, y方向です。

実数の1次元ベクトルなら、内積は「対応する成分を掛け、全部足す」計算です。
`F * s`は成分ごとの積で、まだ内積ではありません。内積の答えは一つの数（スカラー）です。
`F @ s`と`np.dot(F, s)`は、この実数1次元の場合には同じ内積を返します。

````python
import numpy as np

F = np.array([2, 1])       # 力 [N]
s = np.array([3, 0])       # 変位 [m]
print("成分ごとの積 =", F * s)
print("手で書いた内積 [J] =", F[0] * s[0] + F[1] * s[1])
print("@による内積 [J] =", F @ s)
print("dotによる内積 [J] =", np.dot(F, s))
````

## Q9｜行列積は「各行とベクトルの内積」［15分］

### 完成例：二つの入力から二つの出力を作る

ここでは単位を持たない簡単な変換 `y = A x` を考えます。

```
A = [[1, 2],     x = [1, -1]
     [3, 4]]
yの1成分目 = 1×1 + 2×(-1) = -1
yの2成分目 = 3×1 + 4×(-1) = -1
```

`A @ x`の結果は`[-1, -1]`。`A * x`は別の計算で、今回は2行2列になります。
NumPyがxを各行に適用する仕組みを**ブロードキャスト**と呼びます。詳しい規則は別冊で扱います。

**形の確認**：`(m, n) @ (n,) → (m,)`。Aの列数とxの要素数が一致する必要があります。
`(m, n) @ (n, p) → (m, p)`なら行列同士の積です。

````python
import numpy as np

A = np.array([[1, 2], [3, 4]])
x = np.array([1, -1])
print("A @ x =", A @ x)
print("A * x =\n", A * x)
print("@のshape =", (A @ x).shape)
print("*のshape =", (A * x).shape)

I = np.eye(2)                  # 2行2列の単位行列
print("A @ I =\n", A @ I)     # 行列積ではAがそのまま残る
print("A * I =\n", A * I)     # 各要素の積では対角以外が0になる
````

## Q11｜固有値は「ある方向を何倍するか」［20分］

行列を掛けると、ベクトルの向きや長さが変わります。その中には、**同じ直線上にとどまる特別な方向**があります。
0ではないそのベクトルvを固有ベクトル、倍率λを固有値と呼び、`A @ v = λ * v`で確認します。

### 完成例：x方向は2倍、y方向は0.5倍

`A=[[2,0],[0,0.5]]`なら、横向き`[1,0]`は`[2,0]`へ、縦向き`[0,1]`は`[0,0.5]`へ変わります。
この二つの方向の固有値は2と0.5です。

- `np.linalg.eig(A)`は固有値と固有ベクトルを計算します。
- 固有ベクトルは返された配列の**列**に並びます。`vectors[:, i]`は「全ての行の、i列目」です。
- `A@v - λ*v`がほぼ0なら、その組が式を満たすと確認できます。
- `1e-15`はとても小さな数です。小数計算では、厳密な0の代わりに微小な丸め誤差が出ることがあります。

今日は`eig`の中の計算法や、手で固有方程式を解く方法を暗記しません。

````python
import numpy as np

A_eig = np.array([[2.0, 0.0], [0.0, 0.5]])
values, vectors = np.linalg.eig(A_eig)
print("固有値 =", values)
for i in range(len(values)):
    lam = values[i]
    v = vectors[:, i]
    print("λ =", lam, "v =", v)
    print("A @ v =", A_eig @ v)
    print("λ * v =", lam * v)
    print("差 =", A_eig @ v - lam * v)
````

### 図で確かめる

青の実線は元のv、橙の破線はA@vです。矢印の座標と長さを比べます。
**描画の細かい命令は読まなくて構いません。** 最初の行列を入力として図が変わることを見ます。
二つの図は同じ縮尺、xとyも同じ縮尺です。ここでは単位のない数学的な変換を描いています。

````python
import numpy as np
import matplotlib.pyplot as plt

A_plot = np.array([[2.0, 0.0], [0.0, 0.5]])  # 確認問題で左上の2.0を-1.0へ
basis = [np.array([1., 0.]), np.array([0., 1.])]
fig, axes = plt.subplots(1, 2, figsize=(9, 4), layout="constrained")
for ax, v, title in zip(axes, basis, ["x direction", "y direction"]):
    w = A_plot @ v
    ax.axhline(0, color="#808080", linewidth=0.7)
    ax.axvline(0, color="#808080", linewidth=0.7)
    ax.annotate("", xy=v, xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color="#2860A4", lw=3))
    ax.annotate("", xy=w, xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color="#C16C21", lw=2, linestyle="--"))
    ax.plot([], [], color="#2860A4", lw=3, label="v (before)")
    ax.plot([], [], color="#C16C21", lw=2, ls="--", label="A @ v (after)")
    ax.set(xlim=(-1.5, 2.5), ylim=(-1, 2), xlabel="x", ylabel="y",
           title=f"{title}: {v.tolist()} -> {w.tolist()}")
    ax.set_aspect("equal")
    ax.legend(loc="upper left", fontsize=9)
plt.show()
````
