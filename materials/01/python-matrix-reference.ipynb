## R2｜for・if・関数・初期化を一つの例で読む

関数の入力から出力まで追います。`enumerate`は位置と値を同時に取り出します。
`append`はリストの最後に追加します。`return`で判定された位置を返します。
字下げが処理範囲を決め、for内のifで条件に合うデータだけを選びます。

```python
def outside_indices(angles, limit):
    found = []
    for index, angle in enumerate(angles):
        if abs(angle) > limit:
            found.append(index)
    return found

angles = [-8, -3, 0, 4, 7]
print("範囲外の位置 =", outside_indices(angles, 5))
print("もう一度 =", outside_indices(angles, 5))

# 診断用：0へ戻す場所を間違えた例
count_bad = 0
for value in [2, 4, 6]:
    count_bad = 0
    count_bad = count_bad + 1
    print("途中 =", value, count_bad)
print("数えた個数（誤り） =", count_bad)
```

## R4｜内積のいろいろな書き方を比較する

**対象は実数・同じ長さの1次元配列に限定します。** この範囲では下の7通りが同じ内積を計算します。
全てを暗記する必要はありません。まず`@`か`np.dot`を読めるようにし、forで中身を確認します。

|書き方|何をしているか|
|---|---|
|forループ|成分を一組ずつ掛けて加算|
|Pythonのsumとzip|二つの並びを組にし、積を合計|
|np.sum(a*b)|各要素の積を作り合計|
|np.dot(a,b)|1次元では内積|
|np.inner(a,b)|1次元では内積|
|a @ b|1次元同士では内積|
|np.matmul(a,b)|この場合は@と同じ|

配列が2次元以上になるとdot・inner・matmulを単純に同じとは扱えません。
複素数の内積では共役も必要になるため、ここでの実数用の説明をそのまま拡張しません。

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
total = 0
for i in range(len(a)):
    total = total + a[i] * b[i]
methods = {
    "for": total,
    "sum + zip": sum(x * y for x, y in zip(a, b)),
    "np.sum(a*b)": np.sum(a * b),
    "np.dot": np.dot(a, b),
    "np.inner": np.inner(a, b),
    "@": a @ b,
    "np.matmul": np.matmul(a, b),
}
print("成分ごとの積 =", a * b)
for name, result in methods.items():
    print(name, "=", result)
```

## R6｜対角でない行列の固有値

`[[2,1],[1,2]]`では、[1,1]方向は3倍、[1,−1]方向は1倍になります。
計算された固有ベクトルは長さ1へ正規化されていることがあります。符号が逆でも同じ固有方向です。
固有値の順番に頼らず、対応する列のベクトルとセットで検算します。
`np.linalg.norm`はベクトルの長さです。差の長さでずれの大きさを確認します。

```python
import numpy as np

A = np.array([[2., 1.], [1., 2.]])
values, vectors = np.linalg.eig(A)
print("固有値 =", values)
for i, lam in enumerate(values):
    v = vectors[:, i]
    residual = np.linalg.norm(A @ v - lam * v)
    print("λ / v / 差の長さ =", lam, v, residual)
print("[1,1]の変換 =", A @ np.array([1., 1.]))
print("[1,-1]の変換 =", A @ np.array([1., -1.]))
```

## R7｜連立方程式・逆行列・行列式・rank

`A@x=b`を解くには`np.linalg.solve(A,b)`を使います。
逆行列`inv(A)`は学習上比較しますが、連立方程式の計算では通常solveを使います。
`det`は行列式、`matrix_rank`は独立な列（行）の数です。
正方行列でも逆行列が存在しない場合があります。行列式が数値的に小さいかだけで、計算の信頼性全ては判断できません。

```python
import numpy as np

A = np.array([[2., 1.], [1., 3.]])
b = np.array([5., 5.])
x = np.linalg.solve(A, b)
print("solveのx =", x)
print("A@x - b =", A @ x - b)
print("逆行列を使う比較 =", np.linalg.inv(A) @ b)
print("det / rank =", np.linalg.det(A), np.linalg.matrix_rank(A))

singular = np.array([[1., 2.], [2., 4.]])
print("singularのrank =", np.linalg.matrix_rank(singular))
try:
    np.linalg.solve(singular, [3., 6.])
except np.linalg.LinAlgError as error:
    print("この例では一意に解けない:", error)
```

## R8｜norm・trace・累乗・外積を区別する

`norm(v)`は既定でユークリッド長、`trace(A)`は対角成分の和です。
`A**2`は各成分を2乗、`matrix_power(A,2)`はA@Aです。
`outer`（外積・直積と呼ばれる）は全ての成分の組の積を作る行列です。
3次元の`cross`（ベクトル積・外積）は両ベクトルに垂直なベクトルを作ります。同じ「外積」という訳語でも別の演算です。

```python
import numpy as np

v = np.array([3., 4.])
A = np.array([[1, 2], [3, 4]])
print("vの長さ =", np.linalg.norm(v))
print("trace =", np.trace(A))
print("A**2 =\n", A ** 2)
print("A@A =\n", np.linalg.matrix_power(A, 2))
print("outer =\n", np.outer([1, 2], [3, 4]))
print("cross =", np.cross([1, 0, 0], [0, 1, 0]))
```

## R9｜データファイルの読み書きを小さく体験する

CSVは値をカンマで区切ったテキストです。ここでは架空データから作ったCSVを文字列として読み込みます。
`StringIO`でファイルのように扱うため、外部ファイルの準備・アップロードは不要です。
実際のファイルを扱う場合も、列名・単位・欠損値・型を確認する必要があります。

```python
import io
import numpy as np

csv_text = "time_s,angle_deg\n0,-2\n1,0\n2,3\n"
data = np.loadtxt(io.StringIO(csv_text), delimiter=",", skiprows=1)
time_s = data[:, 0]
angle_deg = data[:, 1]
print("shape =", data.shape)
print("時刻 [s] =", time_s)
print("角度 [deg] =", angle_deg)
buffer = io.StringIO()
np.savetxt(buffer, data, delimiter=",", header="time_s,angle_deg", comments="", fmt="%.1f")
print("CSVへ書き出した内容:\n", buffer.getvalue())
```
