## 1｜関数の復習［140–145分］

### 完成例

`count_outside(angles, limit)`は角度の並びとしきい値を受け取り、範囲外の個数を返します。
`count=0`で数え始め、forで各角度を調べ、`count += 1`で一つ増やします（`count=count+1`と同じ）。
この例では−6と8が±5°を超えるので答えは2です。±limitと等しい値は数えません。
`return`は値を呼び出し側へ返し、`print`は画面へ表示します。`assert`は条件が満たされないとエラーで知らせる確認です。

````python
def count_outside(angles, limit):
    count = 0
    for angle in angles:
        if abs(angle) > limit:
            count += 1
    return count
angles = [0, 3, -6, 5, 8, -5]
print(count_outside(angles, 5))
assert count_outside([5,-5,6,-6], 5) == 2
assert count_outside([],5) == 0
assert count_outside(angles,5) == 2
````

## 2｜配列の復習［145–150分］

### 完成例：条件を決めて架空データを作る

これは合成の練習データです。制御器の実験結果でも、実測値でもありません。
`np.linspace(0,6,121)`は0〜6秒を端点を含め121点に分けます。間隔は6/(121−1)=0.05秒です。
`cos(3*time_s)`が振動を作り、`exp(-0.35*time_s)`が時間とともに小さくなる倍率を作ります。
cosの入力はラジアンで、ここでの3は角周波数[rad/s]です。AとBでは小さくなる倍率だけを変えています。
まず初期値と長さを確認します。式の導出や関数名の暗記は求めません。

````python
time_s = np.linspace(0,6,121)
angle_A_deg = 8*np.exp(-0.35*time_s)*np.cos(3*time_s)
angle_B_deg = 8*np.exp(-0.7*time_s)*np.cos(3*time_s)
limit_deg = 5
print("shape:",time_s.shape,"A outside:",count_outside(angle_A_deg,limit_deg))
assert len(time_s)==len(angle_A_deg)==len(angle_B_deg)
````
